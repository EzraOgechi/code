```python
def myFunc():
    x="Jane"
    def myFunc2():
        nonlocal x
        x="Hello"
    myFunc2()
    return x
print(myFunc())  # This will output: Hello
```

```python
def changeCase(func):
    def myInner():
        return func().upper()
    return myInner
@changeCase
def myFunction():
    return "hello world"
print(myFunction())  # This will output: HELLO WORLD
```

```python
def countdown(n):
    if n<=0:
        print("Happy New Year!")
    else:
        print(n)
        countdown(n-1)
countdown(5)
```

```python
def factorial(n):
    if n==0 or  n==1:
        return 1
    else:
        return n * factorial(n-1)
    print(factorial(5))  # This will output: 120
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
for value in my_generator():
    print(value)  # This will output: 1 2 3 on separate lines
```
