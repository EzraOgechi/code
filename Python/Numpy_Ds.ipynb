Universal Functions

```python
import numpy as np
x=[1,2,3,4]
y=[4,5,6,7]
z=np.add(x,y)
print(z)
```

```python
import numpy as np
def myadd(x,y):
    return x+y
myadd=np.frompyfunc(myadd,2,1)
print(myadd([1,2,3,4],[5,6,7,8]))
```

```python
import numpy as np
if type(np.add)==np.ufunc:
    print("add is a ufunc")
else:
    print("add is not a ufunc")
```

Simple Arithmetic

Addition

```python
import numpy as np
arr1=np.array([10,11,12,13,14,15])
arr2=np.array([20,21,22,23,24,25])
newarr=np.add(arr1,arr2)
print(newarr)
```

Subtraction

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,21,22,23,24,25])
newarr=np.subtract(arr1,arr2)
print(newarr)
```

Multplication

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,21,22,23,24,25])
newarr=np.multiply(arr1,arr2)
print(newarr)
```

Division

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,21,22,23,24,25])
newarr=np.divide(arr1,arr2)
print(newarr)
```

Power

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,21,22,23,24,25])
newarr=np.power(arr1,arr2)
print(newarr)
```

Remainder

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,21,22,23,24,25])
newarr=np.mod(arr1,arr2)
print(newarr)
```

Quatient And Mod

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,21,22,23,24,25])
newarr=np.divmod(arr1,arr2)
print(newarr)
```

Absolute Values

```python
import numpy as np
arr1=np.array([-10,20,-30,40,-50,60])
newarr=np.absolute(arr1)
print(newarr)
```

Rounding Decimals

Truncation

```python
import numpy as np
arr=np.trunc([-3.166,3.66667])
print(arr)
```

Rounding

```python
import numpy as np
arr=np.around(3.1666,2)
print(arr)
```

Floor

```python
import numpy as np
arr=np.floor([-3.1666,3.6667])
print(arr)
```

Ceil

```python
import numpy as np
arr=np.ceil([-3.1666,3.6667])
print(arr)
```
