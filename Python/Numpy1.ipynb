```python
import cv2
import numpy as np

# Matrices
protanopia = np.array([
    [0.56667, 0.43333, 0.00000],
    [0.55833, 0.44167, 0.00000],
    [0.00000, 0.24167, 0.75833]
])

deuteranopia = np.array([
    [0.625, 0.375, 0.000],
    [0.700, 0.300, 0.000],
    [0.000, 0.300, 0.700]
])

tritanopia = np.array([
    [0.950, 0.050, 0.000],
    [0.000, 0.433, 0.567],
    [0.000, 0.475, 0.525]
])

def simulate(image, matrix):
    # cv2 uses BGR so convert to RGB
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    h, w, _ = rgb.shape
    flat = rgb.reshape(-1, 3)

    transformed = np.dot(flat, matrix.T)
    transformed = np.clip(transformed, 0, 255)

    result = transformed.reshape(h, w, 3).astype("uint8")
    return cv2.cvtColor(result, cv2.COLOR_RGB2BGR)

# Load image
img = cv2.imread("image.jpg")

# Apply filters
prot = simulate(img, protanopia)
deut = simulate(img, deuteranopia)
trit = simulate(img, tritanopia)

# Show results
cv2.imshow("Original", img)
cv2.imshow("Protanopia", prot)
cv2.imshow("Deuteranopia", deut)
cv2.imshow("Tritanopia", trit)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np
import matplotlib.pyplot as plt
fruits=np.array([20,30,40,10])
colors=['yellow','red','blue','green']
myLabels=['apple','banana','mango','grape']

myExplode=[0.1,0,0,0]

plt.pie(fruits,startangle=90,shadow=True,explode=myExplode,autopct='%1.1f%%', colors=colors,labels=myLabels)
plt.legend(title="Fruits")
plt.show()
```
