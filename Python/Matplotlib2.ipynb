```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the image
image = cv2.imread("image.jpg")          
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 2. Convert to Grayscale
gray = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)

# 3. Apply a color filter: increase saturation
hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)
hsv[:, :, 1] = np.clip(hsv[:, :, 1] + 50, 0, 255)   # Increase saturation
filtered = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

# 4. Display before & after
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.title("Original RGB")
plt.imshow(image_rgb)
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Grayscale")
plt.imshow(gray, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Filtered (Higher Saturation)")
plt.imshow(filtered)
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the image
image = cv2.imread("image.png")  # Replace with your image filename
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 2. Convert to Grayscale
gray = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)

# 3. Apply a color filter: Hue Shift
hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

# Shift hue by +50 (0-179 in OpenCV HSV)
hsv[:, :, 0] = (hsv[:, :, 0] + 50) % 180

hue_shifted = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

# 4. Display before & after
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.title("Original RGB")
plt.imshow(image_rgb)
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Grayscale")
plt.imshow(gray, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Hue Shifted")
plt.imshow(hue_shifted)
plt.axis("off")

plt.tight_layout()
plt.show()
```
