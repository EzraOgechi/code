```python
#-Relationships Between Data in a Table
#A relationship is just a correlation-how columns are related to each other like  if you add data into the dataframe how will it affect the other columns
#Example
"""
Calories and Duration ,, if you increase Duration you will burn  more calories
Hospitals ,, if there are more birth Deaths in an hospital few patients will come


"""
```

```python
import pandas as pd
df=pd.read_csv('data1.csv')
x=df["Calories"].mode()[0]
df.fillna({"Calories":x},inplace=True)
df["Date"]=pd.to_datetime(df['Date'],format='mixed')
df.dropna(subset='Date',inplace=True)
for x in df.index:
    if df.loc[x,"Duration"]>70:
        df.loc[x,"Duration"]=60
        print(df.drop_duplicates())
```

```python
import pandas as pd
df=pd.read_csv('data1.csv')
x=df["Calories"].mode()[0]
df.fillna({"Calories":x},inplace=True)
df["Date"]=pd.to_datetime(df['Date'],format='mixed')
df.dropna(subset='Date',inplace=True)
for x in df.index:
    if df.loc[x,"Duration"]>70:
        df.loc[x,"Duration"]=60
        #print(df.drop_duplicates())
        print(df.corr())
```

```python
#There are three types of correlations
"""
Perfect Correlation---values like 1.0
Good Correlation--ranges from 0.6->0.9// -0.6 t0 -0.9
Bad Correlation --0.0-0.5//-0.0-0  to  -0.5
"""
```

```python
#Plotting your data to see how it looks like,,get insights of your data
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('data1.csv')
x=df["Calories"].mode()[0]
df.fillna({"Calories":x},inplace=True)
df["Date"]=pd.to_datetime(df['Date'],format='mixed')
df.dropna(subset='Date',inplace=True)
for x in df.index:
    if df.loc[x,"Duration"]>70:
        df.loc[x,"Duration"]=60
df.plot(kind="scatter",x="Duration",y="Calories")
plt.show()
```
